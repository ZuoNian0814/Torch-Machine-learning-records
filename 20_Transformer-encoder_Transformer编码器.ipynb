# **任务20 Transformer Encoder | Transformer 编码器**

Transformer架构是一种复杂神经网络结构，它分为编码器`encoder`和解码器`decoder`，这个任务主要为了体现编码器的功能。

对于encoder我们可以简单理解为一个引入了注意力机制和归一化的深度残差感知机网络。可见它是一个叠了很多buff的网络结构，结合了很多用于解决某种特定问题的结构。它作为一种基于感知机的改良，同样可以替换很多使用了它的任务，但要注意是否有必要，毕竟Transformer性能虽然强大，但计算量也很夸张。

结合先前 **任务19** 的位置编码，它即可应用在需要理解空间信息的场景。

## **Transformer编码器层**

实例化方法 `nn.TransformerEncoderLayer(d_model=input_dim, nhead=head_num, batch_first=True, dropout=True)`，其中 `nhead` 是多头注意力机制的头数。`dropout=True`是先前提到的 `nn.Dropout()` 层，`True`表示在模型中启用。

使用 `nn.TransformerEncoder(encoder_layer, num_layers=num_layers)` 可以将编码器层直接封装成多层Transformer编码器，并非必要，直接拼接也可以做到，这个方法主要为了避免代码冗余和设备不匹配等潜在问题。

```python
import torch
import torch.nn as nn

batch_size = 8
seq_num = 4
# 特征值和嵌入维度都是与d_model匹配的
feature_dim = emb_dim = 16
nhead = 8

num_layers = 2

input_tensor = torch.ones(batch_size, emb_dim)  # [seq_len, batch, features]
input_seq_tensor = torch.ones(batch_size, seq_num, emb_dim)  # [seq_len, batch, features]

encoder_layer = nn.TransformerEncoderLayer(dropout=True, d_model=emb_dim, nhead=nhead, batch_first=True)
encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

print('多层Transformer：')
output_tensor = encoder(input_tensor)
output_seq_tensor = encoder(input_seq_tensor)
print('特征值输入张量：', input_tensor.shape)
print('特征值输出张量：', output_tensor.shape)
print('序列数据输入张量：', input_seq_tensor.shape)
print('序列数据输出张量：', output_seq_tensor.shape)

print('\n直接传入Transformer层也可使用：')
output_tensor = encoder_layer(input_tensor)
output_seq_tensor = encoder_layer(input_seq_tensor)
print('特征值输入张量：', input_tensor.shape)
print('特征值输出张量：', output_tensor.shape)
print('序列数据输入张量：', input_seq_tensor.shape)
print('序列数据输出张量：', output_seq_tensor.shape)
```

可见Transformer编码器的内核就是线性层，适配任何线性层作为主要特征提取方法的任务，但要注意它的计算量较大并且参数量下限较高，尽管有为了避免过拟合的Dropout层，但对于简单任务很较发生过拟合。
